```python
import altair as alt
```

```python
import pandas as pd

penguins = pd.read_csv("../penglings.csv")
```

```python
penguins_clean = penguins.dropna(subset=[
    "flipper_length_mm",
    "body_mass_g",
    "bill_length_mm"
])

species_hover = alt.selection_point(
    fields=["species"],
    on="mouseover",
    bind=alt.LegendBinding(legend="mouseover"),
    clear="mouseout",
    toggle=False,
    empty=True
)

alt.Chart(penguins_clean).mark_circle().encode(
    opacity=alt.when(species_hover)
        .then(alt.value(0.8))
        .otherwise(alt.value(0.1)),
    x=alt.X(
        "flipper_length_mm:Q",
        title="Flipper Length (mm)",
        scale=alt.Scale(domain=[170, 235], nice=False)
    ),
    y=alt.Y(
        "body_mass_g:Q",
        title="Body Mass (g)",
        scale=alt.Scale(domain=[2500, 6500], nice=False)
    ),
    size=alt.Size(
        "bill_length_mm:Q",
        title="Bill Length (mm)"
    ),
    color=alt.Color(
        "species:N",
        title="Species",
        legend=alt.Legend(orient="bottom"),
        scale=alt.Scale(
            domain=["Adelie", "Chinstrap", "Gentoo"],
            range=["#FF8C00", "#A032D2", "#008B8B"]
        )
    )
).add_params(species_hover).interactive()
```
