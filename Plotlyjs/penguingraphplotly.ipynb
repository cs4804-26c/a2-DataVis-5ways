```python
import pandas as pd
import plotly.express as px
import plotly.io as pio

pio.renderers.default = "browser"


df = pd.read_csv("penglings.csv").dropna(
    subset=["species", "flipper_length_mm", "body_mass_g", "bill_length_mm"]
)

fig = px.scatter(
    df,
    x="flipper_length_mm",
    y="body_mass_g",
    color="species",
    size="bill_length_mm",
    opacity=0.8,
    color_discrete_map={
        "Adelie": "#FF0000",
        "Chinstrap": "#00FF00",
        "Gentoo": "#0000FF",
    },
    labels=
    {
        "flipper_length_mm": "Flipper Length (mm)",
        "body_mass_g": "Body Mass (g)",
        "bill_length_mm": "Bill Length (mm)",
        "species": "Species",
    },
    title="Flipper Length VS Body Mass",
)

xmin, xmax = df["flipper_length_mm"].min(), df["flipper_length_mm"].max()
ymin, ymax = df["body_mass_g"].min(), df["body_mass_g"].max()

fig.update_xaxes(range=[xmin - 2, xmax + 2], dtick=10)
fig.update_yaxes(range=[ymin - 100, ymax + 100], dtick=500)

fig.show()
```
